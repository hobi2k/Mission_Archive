- `shared` 폴더에 전달된 `model.pkl`과 `test.csv`를 사용해 추론하고 `result.csv`를 저장합니다.

```python
from pathlib import Path

import joblib
import pandas as pd
```

```python
# 경로 설정
shared_dir = Path('/shared')
model_path = shared_dir / 'model.pkl'
test_path = shared_dir / 'test.csv'
```

```python
# 모델 로드 및 테스트 데이터 확인
model = joblib.load(model_path)
test_df = pd.read_csv(test_path)
test_df.head()
```

```python
test_df.info()
```

```python
# 예측 수행
predictions = model.predict(test_df)

# 예측 결과를 데이터프레임 복사 후 추가
result_df = test_df.copy()
result_df['Performance Index'] = predictions.round().astype(int)
```

```python
# 예측 결과 확인
result_df.head()
```

```python
result_path = shared_dir / 'result.csv'
result_df.to_csv(result_path, index=False)
result_path
```
